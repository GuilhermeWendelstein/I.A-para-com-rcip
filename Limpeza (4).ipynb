```python
# --- 2. CARREGAMENTO ----------------------------------------

df = pd.read_excel("Online Retail.xlsx", engine="openpyxl")

# Mantém uma cópia da base original para comparação
df_original = df.copy()

print("Base carregada com sucesso!")
print("Dimensões iniciais:", df.shape)
```

## 3. Inspeção inicial

Antes da aplicação das transformações, foram verificadas as dimensões da base, os tipos das colunas, os valores ausentes e os registros duplicados.

Também foi verificada a existência de uma variável-alvo categórica. Como o projeto utiliza uma abordagem de recomendação e não possui uma variável-alvo categórica definida, a proporção de classes não se aplica nesta etapa.

```python
# --- 3. INSPEÇÃO INICIAL ------------------------------------

print("Dimensões da base:")
print(df.shape)

print("\nTipos das colunas:")
print(df.dtypes)

print("\nValores ausentes por coluna:")
print(df.isnull().sum())

print("\nQuantidade de duplicatas:")
print(df.duplicated().sum())

print("\nProporção de classes da variável alvo:")
print("Não se aplica, pois o projeto não possui uma variável-alvo categórica.")
```

## 4. Tratamento de valores ausentes

A principal coluna com valores ausentes relevante para o projeto é CustomerID.

Como o objetivo é relacionar as compras ao histórico de cada cliente para posteriormente realizar recomendações personalizadas, registros sem CustomerID não podem ser associados corretamente a um consumidor.

Foi escolhida a remoção desses registros em vez do preenchimento, pois não seria adequado criar artificialmente uma identificação de cliente que não existe na base original.

A coluna Description possui uma pequena quantidade de valores ausentes. Esses registros não foram removidos nesta etapa, pois o StockCode continua disponível para identificação do produto. Caso Description seja utilizada diretamente no modelo, esse tratamento poderá ser reavaliado.

```python
# --- 4. TRATAMENTO DE AUSENTES ------------------------------

linhas_antes = len(df)

# Remove somente registros sem identificação de cliente
df = df.dropna(subset=["CustomerID"]).copy()

linhas_depois = len(df)

print("Tratamento de CustomerID:")
print("Linhas antes:", linhas_antes)
print("Linhas depois:", linhas_depois)
print("Linhas removidas:", linhas_antes - linhas_depois)

print("\nValores ausentes restantes:")
print(df.isnull().sum())
```

## 5. Remoção de registros duplicados

Foram removidos os registros que apresentavam todas as informações exatamente iguais.

Essa transformação evita que o mesmo registro seja considerado mais de uma vez nas análises e no treinamento posterior.

A estratégia escolhida foi remover somente duplicatas exatas. Isso reduz o risco de remover registros diferentes que apenas possuem algumas informações em comum.

É importante destacar que InvoiceNo repetido não significa necessariamente duplicidade, pois uma mesma nota pode possuir vários produtos.

```python
# --- 5. DUPLICIDADES ----------------------------------------

linhas_antes = len(df)

# Remove somente registros completamente iguais
df = df.drop_duplicates().copy()

linhas_depois = len(df)

print("Linhas antes:", linhas_antes)
print("Linhas depois:", linhas_depois)
print("Duplicatas removidas:", linhas_antes - linhas_depois)
```

## 7. Codificação de variáveis categóricas

Nesta etapa não foi realizada a codificação das variáveis categóricas.

## 9. Verificação do tipo da coluna InvoiceDate

Foi verificado o tipo da coluna `InvoiceDate` antes do processamento.

A coluna já estava no formato `datetime64[ns]`, portanto não foi necessária uma conversão. Essa verificação confirmou que os dados estavam adequados para análises temporais.

## 10. Remoção de cancelamentos

Foram removidos os registros cujo InvoiceNo começa com a letra C, padrão utilizado na base para identificar cancelamentos.

A transformação foi realizada porque, nesta etapa, o objetivo é trabalhar com compras válidas para identificar padrões de consumo e gerar recomendações.

A estratégia escolhida foi utilizar o próprio padrão existente no InvoiceNo.

O principal risco é perder informações que poderiam ser utilizadas em uma análise específica de cancelamentos ou devoluções.

```python
# --- 10. REMOÇÃO DE CANCELAMENTOS ---------------------------

linhas_antes = len(df)

df = df[~df["InvoiceNo"].astype(str).str.startswith("C")].copy()

linhas_depois = len(df)

print("Linhas antes:", linhas_antes)
print("Linhas depois:", linhas_depois)
print("Cancelamentos removidos:", linhas_antes - linhas_depois)
```

## 11. Validação de Quantity e UnitPrice

Foram removidos os registros que apresentavam Quantity menor ou igual a zero ou UnitPrice menor ou igual a zero.

Para esta etapa, o objetivo é trabalhar com compras válidas e valores positivos de quantidade e preço.

A estratégia escolhida foi manter somente registros com Quantity > 0 e UnitPrice > 0.

O principal risco é que alguns valores negativos podem representar devoluções ou ajustes legítimos. Esses registros foram removidos porque o foco desta base tratada é identificar padrões de compras realizadas.

```python
# --- 11. VALIDAÇÃO DE QUANTITY E UNITPRICE ------------------

linhas_antes = len(df)

df = df[
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0)
].copy()

linhas_depois = len(df)

print("Linhas antes:", linhas_antes)
print("Linhas depois:", linhas_depois)
print("Linhas removidas:", linhas_antes - linhas_depois)
```

## 12. Verificação final

Após todas as transformações, a base será novamente verificada para confirmar suas dimensões, valores ausentes e duplicatas restantes.

Também será comparada a quantidade de linhas e colunas antes e depois do processo de limpeza.

```python
# --- 12. VERIFICAÇÃO FINAL ----------------------------------

print("========== COMPARAÇÃO ANTES X DEPOIS ==========")

print("Linhas antes:", df_original.shape[0])
print("Linhas depois:", df.shape[0])

print("Colunas antes:", df_original.shape[1])
print("Colunas depois:", df.shape[1])

print("\n========== BASE FINAL ==========")

print("Dimensões finais:", df.shape)

print("\nValores ausentes restantes:")
print(df.isnull().sum())

print("\nDuplicatas restantes:")
print(df.duplicated().sum())

print("\nPrimeiras linhas:")
display(df.head())
```
